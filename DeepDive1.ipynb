# Side Effect of Mutability

```python
l = [1,2,3]
l1 = l
#some developer make the copy of existing list and make changes then changes may reflect on first one also. for bypassing this we can use cloning of the list.

# cloning
l1 = l[:]
print(id(l1))
print(id(l))
```

```python
a = (1,2,3,[4,5])
print(id(a))
print(a)
a[-1][-1] = 500
print(id(a))
print(a)
```

```python
a = [1,2,3,(4,5)]
print(id(a))
print(a)

# not possible this
a[-1][-1] = 500
print(id(a))
print(a)
```

```python
# if we perform append/extend on list then there is no change in memory locations
l = [1,2,3]
print(id(l))
print(l)
l.append(4)
print(id(l))
print(l)
```

```python
# but if we concatenate two list then problem arises means memory locations will change
l = [1,2,3]
print(id(l))
print(l)

l = l + [45,54]
print(id(l))
print(l)
```

# Recursion

```python
# you have to multiply two number without using multiply operator
def multiply(a, b, i):
#     base case
    if i == b:
        return 0
    a += multiply(a, b, i+1)
    return a

print(multiply(2,4,0))
    
    
```

```python
def fact(n):
#     base case
    if n == 0 or n == 1:
        return 1
    return n*fact(n-1)

fact(5)
```

```python
#palindrome checker
def palindrome(s, start, end):
#     base case
    if(start > end):
        return True
    if(s[start] != s[end]):
        return False
    return palindrome(s, start+1, end-1)

s = "abba"
print(palindrome(s, 0, len(s)-1))
    
    
```

# higher order function

```python
# you are given a list and you have to print three sum first sum is, sum of even numbers, second is, sum of odd numbers and third sum is sum of those number which are divisible by 3 as well

def result_sum(func, L):
    result = 0
    
    for i in L:
        if func(i):
            result = result + i
    return result

L = [11, 14, 21, 23, 56, 78, 45, 29, 28]

x = lambda x: x % 2 == 0
y = lambda x: x % 2 != 0
z = lambda x: x % 3 == 0

print(result_sum(x, L))
print(result_sum(y, L))
print(result_sum(z, L))
```

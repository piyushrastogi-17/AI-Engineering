# Mutability

```python
# Mutability refers to the ability to change or edit data in it's memory location (List , Dictionary, sets are mutable)


```

```python
a = "hello"
print(id(a))
print(a)

a = a + "world"
print(id(a))
print(a)
```

```python
l = [1, 3, 4]
print(id(l))
print(l)
l[1] = 80
print(id(l))
print(l)
```

# Side Effect of Mutability

```python
l = [1,2,3]
l1 = l
#some developer make the copy of existing list and make changes then changes may reflect on first one also. for bypassing this we can use cloning of the list.

# cloning
l1 = l[:]
print(id(l1))
print(id(l))
```

```python
a = (1,2,3,[4,5])
print(id(a))
print(a)
a[-1][-1] = 500
print(id(a))
print(a)
```

```python
a = [1,2,3,(4,5)]
print(id(a))
print(a)

# not possible this
a[-1][-1] = 500
print(id(a))
print(a)
```

```python
# if we perform append/extend on list then there is no change in memory locations
l = [1,2,3]
print(id(l))
print(l)
l.append(4)
print(id(l))
print(l)
```

```python
# but if we concatenate two list then problem arises means memory locations will change
l = [1,2,3]
print(id(l))
print(l)

l = l + [45,54]
print(id(l))
print(l)
```
